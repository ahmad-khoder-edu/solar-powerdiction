### 2. Theoretical Background

#### 2.1 Linear Regression Theory

Linear regression models the relationship between a dependent variable $y$ and one or more independent variables $X_1, X_2, ..., X_n$ as:

$$y = \beta_0 + \beta_1 X_1 + \beta_2 X_2 + ... + \beta_n X_n + \epsilon$

Where:
- $\beta_0$ is the intercept (bias term)
- $\beta_1, ..., \beta_n$ are the regression coefficients
- $\epsilon$ is the error term (assumed normally distributed)

#### 2.2 Gradient Descent Optimization

Gradient descent is an iterative optimization algorithm used to find the optimal coefficients that minimize the cost function:

**Cost Function (Mean Squared Error):**
$J(\beta) = \frac{1}{2m} \sum_{i=1}^{m} (y_i - \hat{y}_i)^2$

Where:
- $m$ is the number of training examples
- $y_i$ is the actual value
- $\hat{y}_i$ is the predicted value

**Update Rule:**
$\beta_j := \beta_j - \alpha \frac{\partial J}{\partial \beta_j}$

Where:
- $\alpha$ is the learning rate
- $\frac{\partial J}{\partial \beta_j}$ is the partial derivative of the cost function with respect to $\beta_j$

#### 2.3 Regularization

To prevent overfitting, we use L2 regularization (Ridge Regression):

$J_{regularized}(\beta) = \frac{1}{2m} \sum_{i=1}^{m} (y_i - \hat{y}_i)^2 + \lambda \sum_{j=1}^{n} \beta_j^2$

Where $\lambda$ is the regularization parameter.

### 3. Methodology

#### 3.1 Data Collection

**Primary Sources:**
- **PVGIS API:** Historical solar radiation data for Lebanon
  - Global Horizontal Irradiance (GHI)
  - Temperature measurements
  - Sunshine duration
- **Solcast API:** Weather forecasts and historical data
  - Cloud opacity
  - Temperature
  - Optional: Inverter performance logs

**Data Coverage:**
- **Temporal Range:** 2020-2023 (4 years of historical data)
- **Geographic Focus:** Lebanon (Beirut: 33.8938°N, 35.5018°E)
- **Resolution:** Daily aggregated data for model training

#### 3.2 Feature Engineering

**Temporal Features:**
- **Cyclical Encoding:** Day of year and month using sin/cos transformations
- **Seasonal Indicators:** Binary flags for summer, winter, spring, autumn
- **Time-based Features:** Day of week, week of year, quarter

**Weather Features:**
- **Temperature Effects:** Linear and polynomial terms
- **Cloud Coverage:** Direct and interaction terms
- **Derived Variables:** Temperature-solar interactions

**System Features:**
- **Optimal Tilt Angle:** Calculated from PVGIS API
- **Battery Specifications:** Capacity, type, efficiency (if available)
- **Load Profiles:** Daily consumption patterns

#### 3.3 Model Development

**Algorithm Selection:**
- **Primary:** Custom Multiple Linear Regression with Gradient Descent
- **Comparison:** Scikit-learn SGDRegressor for validation
- **Regularization:** L2 regularization to prevent overfitting

**Training Process:**
1. Data preprocessing and scaling
2. Train/test split (80/20)
3. Feature scaling using StandardScaler
4. Model training with cross-validation
5. Hyperparameter tuning (learning rate, regularization strength)

#### 3.4 Evaluation Metrics

**Primary Metrics:**
- **R² (Coefficient of Determination):** Explains variance proportion
- **RMSE (Root Mean Squared Error):** Average prediction error
- **MAE (Mean Absolute Error):** Robust error measure
- **MAPE (Mean Absolute Percentage Error):** Relative error measure

**Validation Approach:**
- **Cross-validation:** 5-fold CV on training data
- **Hold-out Test:** 20% of data for final evaluation
- **Learning Curves:** Monitor training progress

#### 3.5 Battery Integration

**Battery Modeling:**
- **State of Charge (SoC) Tracking:** Real-time battery level
- **Charging/Discharging:** Based on solar generation vs load
- **Depth of Discharge (DoD):** 80% for lithium, 50% for lead-acid
- **Efficiency Losses:** Account for round-trip efficiency

**System Optimization:**
- **Self-Consumption Rate:** Percentage of solar energy used locally
- **Self-Sufficiency Rate:** Independence from grid power
- **Autonomy Days:** System operation without charging
